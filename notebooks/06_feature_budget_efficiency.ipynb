# 06. Top-10 Feature Budget Efficiency
Step 4 통합 근거 순위의 상위 feature를 1개부터 10개까지 누적 적용해 성능, 실행시간, LLM 입력 토큰 footprint를 비교합니다. Random Forest 자체는 토큰을 소비하지 않으며, 토큰 수는 동일 데이터를 LLM 기반 분석기로 전달할 때의 compact JSON 입력량입니다.

```python
from pathlib import Path
import sys

import matplotlib.pyplot as plt
import pandas as pd

PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.insert(0, str(PROJECT_ROOT))
REPORT_DIR = PROJECT_ROOT / 'reports'

from src.secom_analysis import load_secom
from src.modeling import build_baseline_models
from src.feature_budget import cumulative_feature_budget_experiment, feature_budget_conclusions
```

## 1. Top-10 ranking
순위는 Step 4의 통합 근거표 정렬 순서(evidence count → permutation importance → impurity importance)를 그대로 사용합니다.

```python
evidence = pd.read_csv(REPORT_DIR / 'candidate_feature_evidence.csv')
top_features = evidence.head(10)['feature'].tolist()
ranking = pd.DataFrame({'rank': range(1, 11), 'feature': top_features})
ranking.to_csv(REPORT_DIR / 'feature_budget_ranking.csv', index=False)
display(ranking)
```

## 2. Cumulative feature experiment
각 top-k 조합을 Balanced Random Forest로 5-fold × 3회 반복 교차검증합니다. Token은 feature-value JSON만 `cl100k_base`로 측정하며 system/user prompt 같은 고정 overhead는 제외합니다.

```python
X, labels = load_secom(PROJECT_ROOT / 'data/raw')
model = build_baseline_models()['random_forest_balanced']
results = cumulative_feature_budget_experiment(
    model, X, labels['is_fail'], labels['timestamp'], top_features,
    n_splits=5, n_repeats=3, inference_repeats=30
)
results, linearity, optimum = feature_budget_conclusions(results)
results.to_csv(REPORT_DIR / 'feature_budget_comparison.csv', index=False)
linearity.to_csv(REPORT_DIR / 'feature_budget_linearity.csv', index=False)
optimum.to_csv(REPORT_DIR / 'feature_budget_optimum.csv', index=False)
display(results)
display(linearity)
display(optimum)
```

## 3. Performance-cost curves

```python
fig, axes = plt.subplots(2, 2, figsize=(13, 9))
x = results['feature_count']

axes[0, 0].errorbar(x, results['cv_pr_auc_mean'], yerr=results['cv_pr_auc_sem'], marker='o', capsize=3, label='Repeated CV')
axes[0, 0].plot(x, results['temporal_test_pr_auc'], marker='s', label='Temporal Test')
axes[0, 0].set(title='Predictive performance', xlabel='Cumulative top-k features', ylabel='PR-AUC', xticks=x)
axes[0, 0].legend()

axes[0, 1].plot(x, results['fit_time_mean_seconds'], marker='o', label='Fit seconds / fold')
axes[0, 1].plot(x, results['score_time_mean_seconds'], marker='s', label='Score seconds / fold')
axes[0, 1].set(title='Computation time', xlabel='Cumulative top-k features', ylabel='Seconds', xticks=x)
axes[0, 1].legend()

axes[1, 0].plot(x, results['tokens_mean_per_sample'], marker='o', color='#E45756')
axes[1, 0].set(title='LLM input token footprint', xlabel='Cumulative top-k features', ylabel='Mean tokens / sample', xticks=x)

axes[1, 1].scatter(results['tokens_mean_per_sample'], results['cv_pr_auc_mean'], c=x, cmap='viridis', s=70)
for _, row in results.iterrows():
    axes[1, 1].annotate(int(row['feature_count']), (row['tokens_mean_per_sample'], row['cv_pr_auc_mean']), xytext=(4, 3), textcoords='offset points')
axes[1, 1].set(title='Performance vs token budget', xlabel='Mean tokens / sample', ylabel='Repeated-CV PR-AUC')

fig.tight_layout()
fig.savefig(REPORT_DIR / 'feature_budget_efficiency.png', dpi=170)
plt.show()
```

## 4. Conclusions
- 토큰량은 feature 수에 거의 완전 선형으로 증가했습니다(R² 0.9996).
- CV PR-AUC는 전체적으로 상승 추세였지만(R² 0.9174), 5·7·10개 지점에서 정체 또는 소폭 하락해 단조 선형 관계는 아닙니다.
- 학습시간(R² 0.1446)과 추론시간(R² 0.0008)은 feature 수와 선형 관계가 확인되지 않았습니다.
- 평균 CV PR-AUC 최고점과 one-standard-error rule은 모두 상위 9개를 선택했습니다. 10번째 feature는 샘플당 토큰을 76.56에서 85.46으로 늘리지만 CV PR-AUC를 개선하지 못했습니다.
- 따라서 현재 조건의 공식 optimum은 9개입니다. 다만 Temporal PR-AUC는 10개에서 0.0062 높아 rolling backtest로 미래 안정성을 추가 확인해야 합니다.
- Random Forest는 실제 LLM 토큰을 소비하지 않습니다. 토큰값은 소수점 6자리 feature-value compact JSON을 `cl100k_base`로 계산했으며 prompt overhead는 제외했습니다.
