# 01. Data Quality Check
결측률, 상수 feature, 클래스 불균형을 진단하고 보고서 CSV를 생성합니다.

```python
from pathlib import Path
import sys

project_root = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.insert(0, str(project_root))
from src.secom_analysis import load_secom, data_quality_reports
```

```python
X, labels = load_secom(project_root / 'data/raw')
summary, missing, constants, drops, kept = data_quality_reports(X, labels['is_fail'])
display(summary)
display(labels['is_fail'].value_counts().rename(index={0: 'Pass', 1: 'Fail'}))
```

```python
report_dir = project_root / 'reports'
report_dir.mkdir(exist_ok=True)
summary.to_csv(report_dir / 'step1_summary.csv', index=False)
missing.to_csv(report_dir / 'missing_summary.csv', index=False)
constants.to_csv(report_dir / 'constant_features.csv', index=False)
drops.to_csv(report_dir / 'drop_candidate_features.csv', index=False)
print(f'후속 분석 feature: {len(kept)}개')
```
